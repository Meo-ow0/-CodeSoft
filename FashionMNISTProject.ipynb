```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load the dataset
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Normalize the pixel values to be between 0 and 1
train_images = train_images / 255.0
test_images = test_images / 255.0

# Reshape data to fit the CNN (adding 1 channel for grayscale)
train_images = train_images.reshape((train_images.shape[0], 28, 28, 1))
test_images = test_images.reshape((test_images.shape[0], 28, 28, 1))

print("Data successfully loaded!")
```

```python
model = models.Sequential([
    # Convolution layer to extract image features
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    # Second convolution layer for deeper patterns
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Flatten the 2D images into a 1D line of numbers
    layers.Flatten(),

    # Final classification layers
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2), # Prevents memorization/overfitting
    layers.Dense(10, activation='softmax') # 10 outputs for the 10 clothing types
])

print("Model successfully built!")



# Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train for 5 epochs (fast but effective)
model.fit(train_images, train_labels, epochs=5, validation_data=(test_images, test_labels))

# Final check on unseen test data
print("\n--- Final Evaluation ---")
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
```
